```python
import pandas as pd
import os
from rdkit import Chem
from rdkit.Chem import Draw
from tqdm import tqdm

# Paths
csv_path = "pubchem/train_200k.csv"
image_dir = "pubchem/images"
output_csv = "pubchem/train_50k.csv"

# Ensure image directory exists
os.makedirs(image_dir, exist_ok=True)

# Load data
df = pd.read_csv(csv_path).iloc[12000:62000].copy()

# Render and save images
filepaths = []
for idx, row in tqdm(df.iterrows()):
    pubchem_cid = row["pubchem_cid"]
    smiles = row["SMILES"]
    filepath = f"{image_dir}/{pubchem_cid}.png"
    filepaths.append(filepath)

    if not os.path.exists(filepath):  # Avoid redundant work
        try:
            mol = Chem.MolFromSmiles(smiles)
            if mol is not None:
                Draw.MolToFile(mol, filepath, size=(384, 384))
            else:
                print(f"[{pubchem_cid}] Invalid SMILES: {smiles}")
        except Exception as e:
            print(f"[{pubchem_cid}] Failed to render: {e}")

# Add filepath column
df["file_path"] = filepaths

# Save new CSV
df.to_csv(output_csv, index=False)
print(f"Saved: {output_csv}")
```
